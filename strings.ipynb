```python
#Count the number of nucleotides in a sequence

dna="GGGTTTAACCGTGTACAC"
dna.count('T')

a_count=dna.count('A')
c_count=dna.count('C')
g_count=dna.count('G')
t_count=dna.count('T')

print(a_count)
print(c_count)
print(g_count)
print(t_count)

print("The number of G's is", g_count)

at=a_count+t_count
gc=g_count+c_count
length=len(dna)

at_percent=(at/length)*100
gc_percent=(gc/length)*100

print("The AT percentage is", at_percent)
print("The GC percentage is", gc_percent)
```

```python
#Combine strings/sequences

dna1="AGGA"
dna2="TCTC"
combined_sequence=dna1+dna2
print(combined_sequence)
```

```python
#Exercise
dna_seq="AGTTAGCTAGGAG"
#1. Number of nucleotides present
print(len(dna_seq))
#2. GC percentage
gc_count=dna_seq.count('G')+dna_seq.count('C')
gc_percent=(gc_count/len(dna_seq))*100
print("The GC percentage is", gc_percent)
#3. Number of purines present?
ag_count=dna_seq.count('A')+dna_seq.count('G')
print("The number of purines present is", ag_count)
#4. Number of pyrimidines present?
ct_count=dna_seq.count('C')+dna_seq.count('T')
print("The number of pyrimidines present is", ct_count)
#5. Purine percentage
purine_percent=(ag_count/len(dna_seq))*100
print("The purine percentage is", purine_percent)
```
