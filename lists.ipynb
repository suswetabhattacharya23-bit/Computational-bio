```python
#Number of elements in a list
mylist=['A','G','C','T']
number_of_items=len(mylist)
print(number_of_items)
```

```python
#Indexing and slicing
#first element of list
mylist[0]

mylist[0:3]
```

```python
#Sorting a list
mylist.sort()              #sorts the list in alphabetical order
print(mylist)
```

```python
#Reverse a list
mylist.reverse()           #reverses the order of the list
print(mylist)
```

```python
#Insert "TTA" into first position of the list
mylist.insert(0,"TTA")
print(mylist)
```

```python
#Delete items from a list
#approach 1: by indexing

del mylist[0]              #deletes the first element of the list
print(mylist)

#approach 2: by value
mylist.remove("GT")       #removes the first occurrence of "GT" from the list
print(mylist)
```

```python
#Change an elment to another value in a list (for eg-TTA to GCG)
mylist[0]="GCG"
print(mylist)
```
