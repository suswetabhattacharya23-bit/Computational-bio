```python
from Bio import SeqIO
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

input_fasta = "input.fasta"
output_fasta = "just_ids.fasta"

records_with_only_ids = []

# Loop through the original FASTA file
for record in SeqIO.parse(input_fasta, "fasta"):
    # Create a new record keeping only the ID and leaving the sequence blank
    clean_record = SeqRecord(
        Seq(""), 
        id=record.id, 
        description="" # Clears out extra description text if you only want the raw ID
    )
    records_with_only_ids.append(clean_record)

# Save the list to a new FASTA file
SeqIO.write(records_with_only_ids, output_fasta, "fasta")
print(f"Successfully saved headers to {output_fasta}")
```
