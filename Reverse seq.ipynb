```python
#CASE 1: When sequence is a string
#Use slicing

DNA1="ATCTAGAGGATATAC"
DNA1[::-1]

reverse_seq=DNA1[::-1]
print(reverse_seq)

print(DNA1)
print(reverse_seq)
```

```python
#CASE 2:When dealing with list data type
#Approach 1:convert list to string and then find reverse using slicing
DNA2=['A','T','C','T','A','G','A','G','G','A','T','A','T','A','C']
forward_seq=''.join(DNA2)
print(forward_seq)

reverse_seq=forward_seq[::-1]
print(reverse_seq)


#Approach 2:Use list slicing and convert to a string
DNA2=['A','T','C','T','A','G','A','G','G','A','T','A','T','A','C']
rev_seq=DNA2[::-1]
print(rev_seq)
rev_seq_final=''.join(rev_seq)
print(rev_seq_final)


#Approach 3:Apply reverse method on the list and then convert list to string
DNA2=['A','T','C','T','A','G','A','G','G','A','T','A','T','A','C']

#make a copy of the DNA2 list
sequence_copy=[]
sequence_copy.extend(DNA2)
print(sequence_copy)

#reverse method on a list
sequence_copy.reverse()
print(sequence_copy)

reverse_sequence=''.join(sequence_copy)
print(reverse_sequence)
```
