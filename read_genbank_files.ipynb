```python
from Bio import SeqIO

# Define the precise file path and format to parse a single record
file_directory='C:/Users/SB/Documents/GENEBANK/1.gb'
genbank_object=SeqIO.read(file_directory,'gb')
record_id=genbank_object.id
print(record_id)

# Extract the short, internal identifier name designated for this sequence
record_name=genbank_object.name
print(record_name)

# Access the raw nucleotide or protein sequence object
record_sequence=genbank_object.seq
# Calculate the total number of characters (bases) in the sequence string   
sequence_length=len(record_sequence)
print(sequence_length)

# Retrieve the long human-readable definition or title line from the file
description=genbank_object.description
print(description)

# Pull the metadata dictionary containing overarching details
annotations=genbank_object.annotations
print(annotations)

# Store the master list containing all structural elements and sequence coordinates
features=genbank_object.features
# Extract the visual classification type string from every individual feature
feature_types=[feature.type for feature in features]
# Strip away all duplicate strings to keep only unique category names
feature_types=set(feature_types)
# Convert the collection back into a mutable ordered structure for easy indexing
feature_types=list(feature_types)
print(feature_types)

# Overwrite the variable with a manual reference list of specific target types
feature_types=[ 'rRNA', 'tmRNA', 'tRNA', 'gene', 'ncRNA', 'CDS']
print(feature_types)

# Loop through each unique category of features found in the file
for feature in feature_types:
    allfeatures=[i for i in features if i.type==feature]         # Filter the master list to keep only the features matching the current category
    number_of_features=len(allfeatures)                          # Count the total number of items in this filtered category
    print('%s:%d'%(feature,number_of_features))                  # Print the category name and its total count (Format -> Name:Count)
```
