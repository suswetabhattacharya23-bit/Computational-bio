```python
DNA="AGCCCATTGACTG"
```

```python
#INDEXING

DNA[4]                 #Indexing:-Indexing is used to access individual characters of a string. The index of the first character is 0, the second character is 1, and so on. The index of the last character is len(string)-1.
DNA[0]
DNA[12]
DNA[len(DNA)-1]        #Negative indexing:-1 starts from the end of the string;it continues in backward direction

DNA[-1]==DNA[12]
```

```python
#ACCESSING SUBSEQUENCES USING SLICING

DNA[5:9]             #in this splicing,the first index is present in the output but the last index is not present in the output. The output will be from 5 to 8.
DNA[0:4]
DNA[-13:-10]==DNA[0:3]

DNA[5:9:2]           #The third index is called step size. It is used to skip characters in the string. The default step size is 1. If the step size is 2, it will skip every second character in the string.
DNA[-12:-4:-2]       #Python needs positive values for negative strides;otherwise it will return an empty string. The output will be an empty string because the start index is less than the end index and the step size is negative.
```

```python
#SHORT HAND INDEXING ["AGCCCATTGACTG"]

DNA[:10]            #gives results from the start of the string to the 10th index(i.e 9th character)
DNA[:-4]            #gives results from the start of the string to the 4th index from the end(i.e 9th character)
DNA[10:]            #gives results from the 10th index(i.e 11th character) to the end of the string
DNA[2:]
DNA[:]
DNA[::]
```
